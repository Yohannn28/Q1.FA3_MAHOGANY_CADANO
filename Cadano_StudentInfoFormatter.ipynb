```python
# Student Information Formatting Tool

# 1. Get Required Inputs
# Using .strip() removes accidental extra spaces before or after the user's input
first_name = input("Enter your first name: ").strip()
last_name = input("Enter your last name: ").strip()
section = input("Enter your section: ").strip()
favorite_subject = input("Enter your favorite subject: ").strip()

# Combine first and last name into one string
full_name = f"{first_name} {last_name}"

# 2. Format the Data using String Methods
full_name_proper = full_name.title()
full_name_upper = full_name.upper()
section_upper = section.upper()
subject_proper = favorite_subject.title()

# Count the number of characters in the full name (including the space)
name_length = len(full_name)

# 3. Display Required Output
print("\n--- FORMATTED STUDENT PROFILE ---")
print(f"Name: {full_name_proper}")
print(f"Name Code:     {full_name_upper}")
print(f"Section:                   {section_upper}")
print(f"Favorite Subject:          {subject_proper}")
print(f"Character Count (Name):    {name_length}")



# No commit changes
```
