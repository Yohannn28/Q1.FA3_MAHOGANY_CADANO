```python
import datetime

# Event Deadline and Date Formatter

# 1. Get Required Inputs
event_title = input("Enter the event title: ").strip()
year = int(input("Enter the year (e.g., 2026): "))
month = int(input("Enter the month (1-12): "))
day = int(input("Enter the day (1-31): "))

# 2. Create a Date Object using the datetime library
deadline_date = datetime.date(year, month, day)

# 3. Format the Data
title_formatted = event_title.upper()
readable_date = deadline_date.strftime("%B %d, %Y")  # e.g., November 25, 2026
weekday_name = deadline_date.strftime("%A")        # e.g., Wednesday

# 4. Display Required Output
print("\n--- CLASS EVENT DEADLINE ---")
print(f"Event Title:   {title_formatted}")
print(f"Deadline Date: {readable_date}")
print(f"Day of Week:   {weekday_name}")
print(f"Details:       Year: {deadline_date.year} | Month: {deadline_date.month} | Day: {deadline_date.day}")



# No commit changes
```
