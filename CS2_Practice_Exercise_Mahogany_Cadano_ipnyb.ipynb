```python
def update_savings(current_total, deposit):
    total = current_total + deposit
    return total
def check_goal(total, target):
  if target > total:
      print("Keep saving!")
  else:
      print("Goal reached!")
current = int(input("Current money: "))
starget = int(input("Target: "))
depos = int(input("Deposit savings: "))
while depos <= 0:

  print("Invalid deposit")
  depos = int(input("Enter a valid number: "))
newtotal = update_savings(current, depos)
print("new total savings:", newtotal)
check_goal(newtotal, starget)
```
